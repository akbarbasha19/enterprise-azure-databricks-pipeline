```python
%sql
CREATE SCHEMA IF NOT EXISTS e_comm_databricks.analytics;
```

```python
%sql
CREATE OR REPLACE VIEW e_comm_databricks.analytics.vw_sales AS

SELECT

    f.order_id,
    f.order_item_id,
    f.product_id,
    f.seller_id,
    f.customer_id,
    f.price,
    f.freight_value,
    f.order_status,
    f.order_purchase_timestamp,

    c.customer_city,
    c.customer_state,

    p.product_category_name,

    s.seller_city,
    s.seller_state

FROM e_comm_databricks.gold.fact_order_items f

LEFT JOIN e_comm_databricks.gold.dim_customers c
    ON f.customer_id = c.customer_id

LEFT JOIN e_comm_databricks.gold.dim_products p
    ON f.product_id = p.product_id

LEFT JOIN e_comm_databricks.gold.dim_sellers s
    ON f.seller_id = s.seller_id
```

```python
%sql
CREATE OR REPLACE VIEW e_comm_databricks.analytics.vw_payments AS

SELECT

    p.order_id,
    p.payment_type,
    p.payment_value,
    o.order_purchase_timestamp,
    o.customer_id

FROM e_comm_databricks.gold.fact_payments p

LEFT JOIN e_comm_databricks.gold.fact_order_items o
    ON p.order_id = o.order_id
```
