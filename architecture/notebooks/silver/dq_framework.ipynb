```python
def apply_dedup(df, dedup_columns):
    return df.dropDuplicates(dedup_columns)
```

```python
def apply_not_null(df, columns):
    for c in columns:
        df = df.filter(col(c).isNotNull())
    return df
```

```python
def dq_metrics(before_count, after_count):
    return {
        "before": before_count,
        "after": after_count,
        "removed": before_count - after_count
    }
```
