Orders

```python
df_orders = (
    spark.read
        .option("header", True)
        .option("inferSchema", True)
        .option("sep",",")
        .csv("/Volumes/e_comm_databricks/default/olist_raw_volume/olist_orders_dataset.csv")
)
```

```python
df_orders.display()
```

```python
df_orders.printSchema()
```

```python
from pyspark.sql.functions import current_timestamp, col, to_date

df_orders_bronze = (
    df_orders
        .withColumn("_ingest_timestamp", current_timestamp())
        .withColumn("_source_file", col("_metadata.file_path"))
        .withColumn("_ingest_date", to_date(current_timestamp()))
)
```

```python
display(df_orders_bronze)
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS e_comm_databricks.bronze;
```

```python
df_orders_bronze.write \
    .format("delta") \
    .mode("append") \
    .saveAsTable("e_comm_databricks.bronze.olist_orders")
```
