```python
%sql
CREATE OR REPLACE VIEW e_comm_databricks.pipeline_analytics.vw_sales AS

SELECT

    f.order_id,
    f.order_item_id,
    f.product_id,
    f.seller_id,
    f.customer_id,
    f.price,
    f.freight_value,
    f.order_status,
    f.order_purchase_timestamp,

    c.customer_city,
    c.customer_state,

    p.product_category_name,

    s.seller_city,
    s.seller_state

FROM e_comm_databricks.pipeline_gold.fact_order_items f

LEFT JOIN e_comm_databricks.pipeline_gold.dim_customers c
    ON f.customer_id = c.customer_id

LEFT JOIN e_comm_databricks.pipeline_gold.dim_products p
    ON f.product_id = p.product_id

LEFT JOIN e_comm_databricks.pipeline_gold.dim_sellers s
    ON f.seller_id = s.seller_id;
```

```python
%sql
CREATE OR REPLACE VIEW e_comm_databricks.pipeline_analytics.vw_payments AS

SELECT

    payment_type,
    SUM(payment_value) AS total_payment_value,
    COUNT(*) AS payment_count

FROM e_comm_databricks.pipeline_gold.fact_payments

GROUP BY payment_type;
```

```python
%sql
CREATE OR REPLACE VIEW e_comm_databricks.pipeline_analytics.vw_daily_revenue AS

SELECT

    DATE(order_purchase_timestamp) AS order_date,
    SUM(price) AS total_revenue

FROM e_comm_databricks.pipeline_gold.fact_order_items

GROUP BY DATE(order_purchase_timestamp);
```
